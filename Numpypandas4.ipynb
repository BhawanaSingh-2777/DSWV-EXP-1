```python
np.random.seed(42)
marks = np.random.randint(20, 101, size=(12, 4))
print(marks)
print("Student averages:", marks.mean(axis=1))
print("Test means:", marks.mean(axis=0))
print("Test std:", marks.std(axis=0))

passed_all = (marks >= 40).all(axis=1)
print("Students passing all four tests:", passed_all.sum())

avg = marks.mean(axis=1)
print("Students with average >= 75:", np.where(avg >= 75)[0] + 1)
print("Test with greatest variation: Test", marks.std(axis=0).argmax() + 1)
```

```python
m = np.arange(21, 46).reshape(5, 5)       # 5x5
center = m[1:4, 1:4]                      # 3x3
t = center.T                              # 3x3
prod = center @ t                         # (3x3) x (3x3) = 3x3
print("Center:\n", center)
print("Transpose:\n", t)
print("Product:\n", prod)
print("Row sums:", prod.sum(axis=1))
print("Maximum:", prod.max())
print("Positions of maximum:\n", np.argwhere(prod == prod.max()))
```

```python
np.random.seed(0)
qty = np.random.randint(5, 30, size=(6, 4))   # 6 days x 4 products
price = np.array([150, 240, 90, 320])
print(qty)

revenue = qty * price
print("Daily revenue:", revenue.sum(axis=1))
print("Revenue per product:", revenue.sum(axis=0))
print("Highest-revenue day: Day", revenue.sum(axis=1).argmax() + 1)

total_qty = qty.sum(axis=0)
print("Total quantity per product:", total_qty)
print("Best-selling product: Product", total_qty.argmax() + 1)

new_price = np.where(total_qty > 100, price * 0.9, price)
print("New prices:", new_price)
print("Original prices:", price)
print("New revenue per product:", (qty * new_price).sum(axis=0))
print("New total revenue:", (qty * new_price).sum())
```

```python
np.random.seed(10)
data = np.random.normal(50, 12, 100)
print("Values outside 30-70:", ((data < 30) | (data > 70)).sum())

cleaned = np.where(data < 30, 30, np.where(data > 70, 70, data))

print("Mean:", data.mean(), cleaned.mean())
print("Variance:", data.var(), cleaned.var())
print("Std:", data.std(), cleaned.std())
print("Min:", data.min(), cleaned.min())
print("Max:", data.max(), cleaned.max())
```

```python
branch_a = pd.Series({'Q1': 120, 'Q2': 145, 'Q4': 180})
branch_b = pd.Series({'Q2': 110, 'Q3': 135, 'Q4': 155})
branch_c = pd.Series({'Q1': 95, 'Q3': 125, 'Q4': 140})

total = branch_a + branch_b + branch_c
print(total)
common = branch_a.index.intersection(branch_b.index).intersection(branch_c.index)
print("Common quarters:", list(common))
print("Combined earnings:\n", total[common])

total0 = branch_a.add(branch_b, fill_value=0).add(branch_c, fill_value=0)
print("Treating absent records as zero:\n", total0)
```

```python
scores = {'R101': 78, 'R102': 65, 'R104': 92, 'R105': 48}
required = ['R105', 'R103', 'R101', 'R106', 'R104', 'R102']
s = pd.Series(scores, index=required)
print(s)
print("Missing records:", list(s[s.isnull()].index))

valid = s[s >= 70]
print(valid)

d = valid.to_dict()
s2 = pd.Series(d)
print(s2)
print("Labelled values match:", s2.equals(valid))
```
