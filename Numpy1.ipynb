```python
marks = np.random.randint(30, 100, size=(10, 3))   # 10 students x 3 subjects
print("Marks:\n", marks)
total = marks.sum(axis=1)
average = marks.mean(axis=1)
print("Total:", total)
print("Average:", average)
print("Students with average >= 50 (index):", np.where(average >= 50)[0])
print("Result:", np.where(average >= 50, 'Pass', 'Fail'))
```

```python
m = np.arange(1, 17).reshape(4, 4)
print("Matrix:\n", m)
print("Transpose:\n", m.T)
print("Row-wise sums:", m.sum(axis=1))
print("Column-wise sums:", m.sum(axis=0))
print("Mean:", m.mean())
print("Maximum:", m.max())
print("Minimum:", m.min())
print("Central 2x2:\n", m[1:3, 1:3])
```

```python
data = np.random.randn(20)
print("Data:", data)
print("Mean:", data.mean())
print("Std deviation:", data.std())
print("Minimum:", data.min())
print("Maximum:", data.max())
print("Sorted:", np.sort(data))
print("Negatives replaced with 0:", np.where(data < 0, 0, data))
```

```python
A = np.random.randint(1, 10, (3, 3))
B = np.random.randint(1, 10, (3, 3))
print("A:\n", A)
print("B:\n", B)
print("Element-wise addition:\n", A + B)
print("Element-wise multiplication:\n", A * B)
prod = np.dot(A, B)
print("Matrix multiplication:\n", prod)
print("Transpose of A:\n", A.T)
print("Max of result:", prod.max())
print("Min of result:", prod.min())
```

```python
a = np.arange(1, 25).reshape(4, 6)
print("Matrix:\n", a)
print("Selected rows 0 and 2:\n", a[[0, 2]])
print("Selected columns 1 and 4:\n", a[:, [1, 4]])
print("Rows 1-2, columns 2-4:\n", a[1:3, 2:5])
print("Transpose:\n", a.T)
print("Row sums:", a.sum(axis=1), "| Row means:", a.mean(axis=1))
print("Column sums:", a.sum(axis=0), "| Column means:", a.mean(axis=0))
print("Unique values:", np.unique(a))
```
